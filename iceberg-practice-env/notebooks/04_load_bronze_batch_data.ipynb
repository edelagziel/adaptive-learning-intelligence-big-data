```python
learner_profiles_path = (
    "/home/iceberg/notebooks/jasonData/learner_profiles.json"
)

learner_profiles_raw_df = (
    spark.read
    .option("multiline", "true")
    .json(learner_profiles_path)
)

learner_profiles_raw_df.show(truncate=False)
learner_profiles_raw_df.printSchema()
```

```python
from pyspark.sql.functions import col, to_timestamp, to_json

learner_profiles_bronze_df = (
    learner_profiles_raw_df
    .select(
        col("user_id"),
        to_timestamp(col("profile_updated_at")).alias("profile_updated_at"),
        to_timestamp(col("ingestion_time")).alias("ingestion_time"),
        col("source_system"),
        to_json(col("raw_payload")).alias("raw_payload")
    )
)

learner_profiles_bronze_df.show(truncate=False)
learner_profiles_bronze_df.printSchema()
```

```python
learner_profiles_bronze_df.writeTo(
    "demo.bronze.learner_profiles"
).append()
```

```python
spark.sql("""
SELECT COUNT(*) AS row_count
FROM demo.bronze.learner_profiles
""").show()
```

```python
spark.sql("""
SELECT *
FROM demo.bronze.learner_profiles
ORDER BY user_id, profile_updated_at
""").show(truncate=False)
```
